CHART 5: (AGE VS PURCHASE AMOUNT SCATTER PLOT FOR AGES 40-50 WITH PURCHASES UNDER 80)

In this chart we filtered the data conditionally to sort it such as customers aged between 40 and 50 were selected. Purchase amounts below 80 were included and stored them in the 'filtered data' dataset. CHART 6: (PURCHASE AMOUNT OVER TIME)

The "Purchase Date" column was converted to a proper date format. This ensures that the dates are correctly interpreted for time-based analysis. The data was grouped by "Purchase Date" to calculate the total "Purchase Amount" for each day. This gave us the sum of all purchases made on each specific date. then we created a line plot. This analysis supports the goal of examining sales trends in the e-commerce dataset over time to identify patterns in purchasing behavior and reveal how purchase amounts change everyday.




```python
df['Purchase Date'] = pd.to_datetime(df['Purchase Date']) #converting to the date time format 
Amount = df.groupby('Purchase Date')['Purchase Amount'].sum()
Amount
```

```python
plt.figure(figsize = (15,7))
plt.plot(Amount.index , Amount.values)
plt.title("Total Purchase Amount Over Time" , fontsize = 17)
plt.xlabel('Date', fontsize = 13)
plt.ylabel('Total Purchase Amount',fontsize = 13)
plt.grid(True)
plt.tight_layout() #to prevent overlapping
plt.show()




```

CHART 7: (AVERAGE PURCHASE AMOUNT BY AGE)

We grouped the data by "Age" and calculated the mean "Purchase Amount" for each age group. This gave us the average amount spent by each age group. Visualized this using a line plot since it is the most suitable choice to visualize such data. This analysis helps us understand how different age groups contribute to purchasing behavior and whether certain age groups are more consistent in their spending.


```python
mean_amount = df.groupby('Age')['Purchase Amount'].mean()
plt.xlabel('Age')
plt.ylabel('Average Purchase Amount')
plt.plot (mean_amount.index,mean_amount.values,marker = "o",color = 'pink')
plt.title("Average Purchase Amount By Age")
plt.show()



```
